```python
import pandas as pd

data = pd.read_csv("E:\Anushka-playground\Gen_AI\Churn_Modelling.csv")
# print(data)
data.head()
```

```python
import pandas
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
import pickle
```

```python
data = data.drop(['RowNumber', 'CustomerId', 'Surname'], axis= 1)
data
```

```python
Encoder = LabelEncoder()
data['Gender'] = Encoder.fit_transform(data['Gender'])
data
```

```python
from sklearn.preprocessing import OneHotEncoder
onehot = OneHotEncoder(sparse_output=False)
onehotgeo = onehot.fit_transform(data[['Geography']])
onehotgeo
```

```python
print(onehot.categories_)
```

```python
onehot.get_feature_names_out(['Geography'])
```

```python
encode_df = pd.DataFrame(onehotgeo, columns=onehot.get_feature_names_out(['Geography']))
encode_df
```

```python
data.head()
```

```python
data = pd.concat([data.drop('Geography' , axis = 1),encode_df], axis=1)
data.head()
```

```python
with open('Encoder.pkl', 'wb') as file:
    pickle.dump(Encoder,file)

with open('onehot.pkl', 'wb') as file:
    pickle.dump(onehot,file)
```

```python
x = data.drop('Exited',axis = 1)
y = data['Exited']

x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)

scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
x_train
```

```python
with open('scaler.pickle', 'wb') as file:
    pickle.dump(scaler,file)
```

STARTING WITH IMPLEMENTATION


```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping, TensorBoard
import datetime
```

```python
(x_train.shape[1],) #form of tupple
```

```python
model = Sequential([
    Dense(64,activation='relu',input_shape = (x_train.shape[1],)), #hidden layer1 with input layer
    Dense(32,activation='relu'), #hidden layer 2
    Dense(1,activation='sigmoid') #output layer
]
)
```

```python
model.summary()
```

```python
import tensorflow
opt = tensorflow.keras.optimizers.Adam(learning_rate=0.001)
loss = tensorflow.keras.losses.BinaryCrossentropy() 
loss
```

```python
model.compile(optimizer=opt, loss="binary_crossentropy", metrics=['accuracy'])
```

```python
from tensorflow.keras.callbacks import EarlyStopping, TensorBoard
log_dir = 'logs/fit' + datetime.datetime.now().strftime("%Y%m%d-%H%M.%S")
tensorflow_callback = TensorBoard(log_dir=log_dir,histogram_freq=1)
```

```python
early_stopping_callback = EarlyStopping(monitor='val_loss',patience=20,restore_best_weights=True)
```

```python
#training
history = model.fit(
    x_train,y_train,validation_data = (x_test,y_test),epochs = 100,
    callbacks =[ tensorflow_callback, early_stopping_callback]
)
```

```python
model.save('model.keras')
```
